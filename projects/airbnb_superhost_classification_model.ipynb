# Airbnb Superhost Classification Model

## Objective
This project aims to build a classification model to predict whether an Airbnb host is likely to become a **Superhost**, based on host performance, listing characteristics, and customer feedback.

## Approach
- Data preprocessing and cleaning  
- Handling class imbalance  
- Exploratory Data Analysis (EDA)  
- Feature engineering  
- Classification modeling (Decision Tree, Random Forest)  
- Model evaluation using cross-validation  

## Key Insights
- Review scores are the strongest predictor of Superhost status  
- Response rate and responsiveness significantly impact host success 
- Consistency in performance is more important than one-time high ratings  
- Random Forest outperforms single models, improving prediction accuracy  

## Model Performance
- Evaluated using classification metrics (Accuracy, Precision, Recall)  
- Cross-validation applied to ensure model robustness  
- Ensemble methods improved overall prediction stability  

## Business Implications
- Helps Airbnb identify high-potential hosts early  
- Enables targeted recommendations to improve host performance  
- Supports platform trust and customer satisfaction  

## Tools & Technologies
- Python (pandas, numpy)  
- scikit-learn  
- Decision Tree & Random Forest  
- Data visualization (matplotlib, seaborn)

### 1.2. Data transformations

#### a. Convert 'amenities' to 'amenities_count'

This will be better instead of dropping this column out of our dataset because
* The number of amenities strongly reflects listing quality --> affects guest satisfaction --> affects review scores --> affects chances of being Superhost.

```python
listing["amenities"]
```

```python
import ast
import pandas as pd

def count_amenities(x):
    # Case 1: already a Python list
    if isinstance(x, list):
        return len(x)
    
    # Case 2: it's a string
    if isinstance(x, str):
        s = x.strip()
        if s == "":
            return 0
        try:
            # Try to interpret it as a Python-style list string
            items = ast.literal_eval(s)
            return len(items)
        except Exception:
            # Fallback: treat it as a comma-separated string
            return len([a for a in s.split(',') if a.strip() != ""])
    
    # Case 3: NaN or anything else
    return 0

listing['amenities_count'] = listing['amenities'].apply(count_amenities)
```

```python
listing["amenities_count"].head()
```

```python
listing.drop(columns=['amenities'], inplace=True)
```

* price: Numeric variables with highly skewed distributions --> replace missing price values using the mean, but only using values within the 3rd quartile (≤ 75th percentile).
* beds: Use mode

```python
listing["price"].describe() #very highly skewed 
```

```python
# 1. Compute Q3 (75th percentile)
q3 = listing["price"].quantile(0.75)

# 2. Compute mean within Q3 (mean of lower 75% of prices)
mean_within_q3 = listing[listing["price"] <= q3]["price"].mean()

# 3. Replace missing values using this trimmed mean
listing["price"] = listing["price"].fillna(mean_within_q3)
```

This will be better instead of dropping this column out of our dataset because
* The number of amenities strongly reflects listing quality --> affects guest satisfaction --> affects review scores --> affects chances of being Superhost.

```python
test["amenities"]
```

```python
import ast
import pandas as pd

def count_amenities(x):
    # Case 1: already a Python list
    if isinstance(x, list):
        return len(x)
    
    # Case 2: it's a string
    if isinstance(x, str):
        s = x.strip()
        if s == "":
            return 0
        try:
            # Try to interpret it as a Python-style list string
            items = ast.literal_eval(s)
            return len(items)
        except Exception:
            # Fallback: treat it as a comma-separated string
            return len([a for a in s.split(',') if a.strip() != ""])
    
    # Case 3: NaN or anything else
    return 0

test['amenities_count'] = test['amenities'].apply(count_amenities)
```
